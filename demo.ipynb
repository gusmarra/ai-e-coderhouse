# Demo: delegación en el orquestador multi-agente

Una consulta que **obliga** a pasar por los dos dominios: el *Investigador* busca la política en los documentos del Hotel Bahía Serena y el *Analista* calcula la penalidad. El *Supervisor* decide quién interviene, valida contra su rúbrica y cierra con la síntesis.

Las salidas de este notebook son las de una corrida real con `gemini-3.5-flash-lite` (guardadas en `trazas/traza_demo.json`).

```python
from graph import construir_grafo
from llm import crear_modelo
from main import CONSULTA_DEMO
from trazas import ejecutar, formatear_evento

grafo = construir_grafo(crear_modelo())
print(CONSULTA_DEMO)
```

## 1. La topología (generada desde el propio grafo)

```python
from IPython.display import Image

Image(grafo.get_graph().draw_mermaid_png())
```

## 2. La ejecución, nodo por nodo

Cada bloque es un turno. El supervisor aparece entre cada especialista: **nunca hablan entre sí**.

```python
traza = await ejecutar(grafo, CONSULTA_DEMO, on_evento=lambda e: print(formatear_evento(e), end=chr(10) * 2))
```

## 3. Quién aportó qué (bitácora del estado compartido)

```python
print(" -> ".join(traza.ruta))
for c in traza.contribuciones:
    print(f"paso {c['paso']} · {c['agente']:<12} · herramientas={c['herramientas']}")
    print(f"    consigna: {c['instruccion']}")
```

## Qué se ve en la demo

1. **Supervisor → Investigador**: consigna concreta, sin cuentas. El investigador usa `buscar_en_documentos_hotel` y devuelve datos con su fuente (`politica-reservas-cancelaciones.md`).
2. **Supervisor → Analista**: valida la rúbrica (fuentes ✓) y delega el cómputo. El analista **no ve la charla**: recibe su consigna y el hallazgo del investigador. Usa `calcular` (no hace la cuenta "de cabeza").
3. **Supervisor → FINISH**: rúbrica completa (fuentes, completitud, cómputo, consistencia) → pasa a `sintesis`.
4. **Síntesis**: redacta la respuesta final citando la fuente.

Si el supervisor pidiera refinamientos sin parar, las guardas (`MAX_PASOS`, `MAX_REFINAMIENTOS`, `recursion_limit`) fuerzan el cierre: ver `tests/test_grafo.py::test_supervisor_infinito_*`.
